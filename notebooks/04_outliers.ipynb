```python
import pandas as pd

clean = pd.read_csv("../data/retail_clean.csv", parse_dates=["InvoiceDate"])
big = clean[clean["Quantity"] > 10000]
big[["Invoice", "InvoiceDate", "Description", "Quantity", "Price", "CustomerID", "Revenue"]]
```

```python
raw = pd.read_csv("../data/retail_raw.csv", parse_dates=["InvoiceDate"])
raw["Invoice"] = raw["Invoice"].astype(str)

cancels = raw[raw["Invoice"].str.startswith("C") & (raw["Quantity"].abs() > 10000)]
cancels[["Invoice", "InvoiceDate", "Description", "Quantity", "Price", "Customer ID"]]
```

```python
before_rows = len(clean)
before_rev = clean["Revenue"].sum()

bad = clean["Invoice"].astype(str).isin(["541431", "581483"]) & (clean["Quantity"] > 10000)
clean = clean[~bad]

clean.to_csv("../data/retail_clean.csv", index=False)

print("rows removed:", before_rows - len(clean))
print("revenue removed:", round(before_rev - clean["Revenue"].sum(), 2))
print("new total revenue:", round(clean["Revenue"].sum(), 2))
```
