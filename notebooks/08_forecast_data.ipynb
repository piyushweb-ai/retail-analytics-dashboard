```python
import pandas as pd
import matplotlib.pyplot as plt

clean = pd.read_csv("../data/retail_clean.csv", parse_dates=["InvoiceDate"])

monthly = clean.set_index("InvoiceDate")["Revenue"].resample("MS").sum().reset_index()
monthly.columns = ["Month", "Revenue"]
monthly.tail(3)
```

```python
monthly = monthly[monthly["Month"] < "2011-12-01"].reset_index(drop=True)
print(len(monthly), "months:", monthly["Month"].min().date(), "->", monthly["Month"].max().date())
```

```python
monthly.set_index("Month")["Revenue"].plot(figsize=(11, 4), marker="o",
                                           title="Monthly revenue (Dec 2009 - Nov 2011)")
plt.ylabel("Revenue")
plt.show()

monthly.to_csv("../data/monthly_revenue.csv", index=False)
```
