```python
import pandas as pd

clean = pd.read_csv("../data/retail_clean.csv", parse_dates=["InvoiceDate"])
cust = clean.dropna(subset=["CustomerID"]).copy()
cust["CustomerID"] = cust["CustomerID"].astype(int)

snapshot = cust["InvoiceDate"].max() + pd.Timedelta(days=1)
print("customers:", cust["CustomerID"].nunique())
print("snapshot date:", snapshot)
```

```python
rfm = cust.groupby("CustomerID").agg(
    Recency=("InvoiceDate", lambda x: (snapshot - x.max()).days),
    Frequency=("Invoice", "nunique"),
    Monetary=("Revenue", "sum"),
).reset_index()

rfm.head()
```

```python
print(rfm.describe().round(1))
rfm.to_csv("../data/rfm.csv", index=False)
```
