```python
import os, shutil
import pandas as pd

os.makedirs("../dashboard_data", exist_ok=True)
clean = pd.read_csv("../data/retail_clean.csv", parse_dates=["InvoiceDate"], low_memory=False)

pd.DataFrame([{
    "total_revenue": clean["Revenue"].sum(),
    "orders": clean["Invoice"].nunique(),
    "customers": clean["CustomerID"].nunique(),
    "start": clean["InvoiceDate"].min().date(),
    "end": clean["InvoiceDate"].max().date(),
}]).to_csv("../dashboard_data/kpis.csv", index=False)

(clean.groupby("Country")["Revenue"].sum().sort_values(ascending=False)
      .reset_index().to_csv("../dashboard_data/country_revenue.csv", index=False))

(clean.groupby("Description")["Revenue"].sum().sort_values(ascending=False).head(15)
      .reset_index().to_csv("../dashboard_data/top_products.csv", index=False))

for f in ["monthly_revenue.csv", "rfm_clusters.csv", "forecast_next_6_months.csv"]:
    shutil.copy(f"../data/{f}", f"../dashboard_data/{f}")

print(sorted(os.listdir("../dashboard_data")))
print(pd.read_csv("../data/rfm_clusters.csv", nrows=1).columns.tolist())
```
